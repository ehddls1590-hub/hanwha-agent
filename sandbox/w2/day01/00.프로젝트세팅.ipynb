## 주피터 노트북 시작하기


```python
import os
import pathlib

here = pathlib.Path.cwd()

ROOT = here.parents[2] if here.name == "day01" else here

os.chdir(ROOT) # 앞으로 모든 상대경로는 이 폴더가 기준이 된다.

SANDBOX = ROOT / "sandbox" / "w2" / "Day01"
# SANDBOX.mkdir(parents=True, exist_ok=True)
```

```python
print(ROOT)
print(SANDBOX.relative_to(ROOT))
```

```python
(SANDBOX / "hello_mod.py").write_text("MESSAGE = '나는 sandbox/w2/day01안에 있다'", encoding="utf-8")
print("만든 파일 : ", (SANDBOX / "hello_mod.py").relative_to(ROOT))
```
